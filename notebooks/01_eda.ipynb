# Credit Card Fraud Detection

## 1. Problem Statement & imports
The dataset contains transactions from European cardholders 
for September 2013. The task is to detect fraudulent transactions.

**Problem:** Only 0.17% of transactions are fraudulent—a severe class imbalance.
**Goal:** Build a model with high recall for the fraud class.

## 3. First Look at the Data

**Dataset shape:** 284,807 transactions × 31 columns

**Key observations:**
- No missing values — expected for a Kaggle competition dataset,
  which has already been preprocessed
- Features V1–V28 are the result of PCA transformation (anonymized
  for confidentiality reasons)
- Only `Time`, `Amount` and `Class` remain in original form
- `Class` is our target: 0 = legitimate, 1 = fraud

**Next step:** check class balance — this is the core challenge
of this dataset

```python
df['Class'].value_counts()
```

## 4. Class Imbalance — The Core Problem

284,315 legitimate vs 492 fraud transactions.
Fraud accounts for only **0.17%** of all transactions.

This means a model that predicts "legitimate" for every transaction
achieves **99.83% accuracy** — but catches zero fraud.
That's why accuracy is a useless metric here. We'll use ROC-AUC and PR-AUC instead.

```python
fig, ax = plt.subplots(figsize=(6, 4))

sns.countplot(x="Class", data=df, hue='Class', palette=["steelblue", "crimson"])

plt.title("Class Distribution: Legitimate vs Fraud")
plt.xlabel("Class (0 = Legitimate, 1 = Fraud)")
plt.ylabel("Count")
plt.show()
```

## 5. Exploring Key Features

V1-V28 are PCA-transformed — we can't interpret them directly.
Let's focus on `Amount` and `Time` — the only original features.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(df["Amount"], bins=50, ax=axes[0])
axes[0].set_title("Transaction Amount Distribution")

sns.histplot(df["Time"], bins=50, ax=axes[1])
axes[1].set_title("Transaction Time Distribution")

plt.tight_layout()
plt.show()
```

## 6. Key EDA Findings & Next Steps

**Findings:**
- Severe class imbalance: 99.83% legitimate vs 0.17% fraud
- `Amount` is right-skewed with outliers up to 25,000
- `Time` reflects daily transaction patterns (2 days of data)
- V1-V28 are already PCA-transformed and roughly scaled

**Preprocessing needed (next notebook):**
